```python
def classify_stock_level(x):
    if x < 50:
        return "Low"
    elif 50 <= x < 80:
        return "Medium"
    else:
        return "High"
df2['Stock_group']=df2['stock_per'].apply(classify_stock_level)
print(df2['Stock_group'])
```

```python
def classify_avail_level(y):
    if y < 50:
        return "Low"
    elif 50 <= y < 80:
        return "Medium"
    else:
        return "High"
df2['Avail_group']=df2['avail_per'].apply(classify_avail_level)
print(df2['Avail_group'])
```

```python
from pandas.api.types import CategoricalDtype
order=["Low", "Medium", "High"]
cat_type=CategoricalDtype(categories=order, ordered=True)
df2['Stock_group']=df2['Stock_group'].astype(cat_type)
ct=pd.crosstab(df2['p_type'], df2['Stock_group'])
ct_percent=pd.crosstab(df2['p_type'], df2['Stock_group'], normalize='index')*100
combined=ct.astype(str) + " (" + ct_percent.round(1).astype(str) + "%) " 
print(combined)
```

```python
order=["Low", "Medium", "High"]
cat_type=CategoricalDtype(categories=order, ordered=True)
df2['Avail_group']=df2['Avail_group'].astype(cat_type)
ct=pd.crosstab(df2['p_type'], df2['Avail_group'])
ct_percent=pd.crosstab(df2['p_type'], df2['Avail_group'], normalize='index')*100
combined=ct.astype(str) + " (" + ct_percent.round(1).astype(str) + "%) " 
print(combined)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ct = pd.crosstab(
    index=df2["Stock_group"],
    columns=df2["Avail_group"],
    values=df2["order_qun"],
    aggfunc="sum",
    normalize=False
)
print(ct)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(ct, annot=True, fmt=".0f", cmap="Blues")
plt.title("Crosstab of Stock_group vs Avail_group (Colored by order_qun)")
plt.xlabel("Stock_group")
plt.ylabel("Avail_group")
plt.show()
```

```python
ct = pd.crosstab(
    index=df2["gender"],
    columns=df2["delv_location"],
    values=df2["no_prod_sold"],
    aggfunc="sum",
    normalize=False
)
print(ct)
```

```python
ct_long = ct.reset_index().melt(
    id_vars="gender",
    var_name="delv_location",
    value_name="no_prod_sold"
)
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=ct_long,
    x="gender",
    y="no_prod_sold",
    hue="delv_location"
)

plt.title("Number of Products Sold by Gender and Delivery Location")
plt.xlabel("Gender")
plt.ylabel("Total Products Sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
def classify_def_level(x):
    if x < 1:
        return "Excellent"
    elif 1 <= x <= 2:
        return "Acceptable"
    else:
        return "Action Required "
df2['AQL']=df2['def_rate'].apply(classify_def_level)
print(df2['AQL'])
```

```python
def classify_inspect_resl_level(x):
    if x =="Pass":
        return 1
    else:
        return 0
df2['Pass']=df2['inspect_resl'].apply(classify_inspect_resl_level)
print(df2['Pass'])
```

```python
Pass_freq = df2["Pass"].value_counts().reset_index()
```
